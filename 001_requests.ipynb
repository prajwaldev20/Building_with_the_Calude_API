```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Make a request

try:
    message = client.messages.create(
        model = model,
        max_tokens=1000,
        messages=[
            {
                "role":"user",
                "content": "what is quantum computing? Answer in one simple sentence"
            }
        ]
    )

except Exception as e:
    print(type(e))
    print(e)
```
